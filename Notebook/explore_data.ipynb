```python
from pyspark.sql import SparkSession

# Connect to Spark with HDFS
spark = SparkSession.builder \
    .appName("Explore HDFS Data") \
    .master("local[*]") \
    .config("spark.hadoop.fs.defaultFS", "hdfs://namenode:9000") \
    .getOrCreate()

# Read enriched Snowplow data
df = spark.read.text("/output")

df.show(truncate=False)
```

```python
from pyspark.sql.functions import from_json, col
from pyspark.sql.types import StructType, StringType

schema = StructType().add("event", StringType())  # Extend if known structure

df_parsed = df.select(from_json(col("value"), schema).alias("data"))
df_parsed.select("data.*").show()
```
